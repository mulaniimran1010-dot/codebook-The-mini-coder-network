```python
import json

def clean_data(data):
    
    #removes user with missing value name
    data["users"]=[user for user in data["users"]if  user["name"].strip()]

    #removes duplicate friend
    for users in data["users"] :  
        users['friends']= list(set(users["friends"]))

    #removes inactive users
    data["users"]=[user for user in data["users"] if user["friends"] or user["liked_pages"]]

    # removes duplicate page
    unique_pages= {} 
    for page in data ["pages"]:
       unique_pages[page["id"]]= page
    data['pages'] = list (unique_pages.values())

    return data
    
data=json.load(open("data5.json"))
data= clean_data(data)
json.dump(data,open("cleaned_data2.json","w"), indent=4)
print("the data is cleaned, you are ready for 10+lpa job ")

```
